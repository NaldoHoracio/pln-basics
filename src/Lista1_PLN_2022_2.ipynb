```python
def is_valid_cpf(cpf_candidate):
  cpf_pattern = "(\d{3}).?(\d{3}).?(\d{3})-?(\d{2})"
  search = re.compile(cpf_pattern).search(cpf_candidate)

  if not search:
    return "NO MATCH: invalid CPF.\n"
  else:
    return "MATCH: VALID CPF.\n"
```

```python
def is_valid_address(address_candidate):
  pattern = "([A-Z]{1}\w+\s*\w*\s*\w*\s*\w*)[,]{1}"\
  "\s*(\d+)[,]{1}"\
  "\s?(ap{1}t*\s+\d+[,]{1})?"\
  "\s?([A-Z]{1}\w+\s*\w*\s*\w*\s*\w*[,]{1})?"\
  "\s?(\w{3}\s+\d{5}-+\d{3})[.]{1}"\
  "\s?([A-Z]{1}\w+\s*\w*\s*\w*\s*\w*)[,]{1}"\
  "\s?([A-Z]{2})[.]{1}"
  compile_pattern = re.compile(pattern)
  addr = compile_pattern.match(address_candidate)

  if not addr:
    return "NO MATCH! Address on INVALID pattern!"
  else:
    print("MATCH! Address VALID pattern! See the address:")
    print("Rua: ", addr.group(1))
    print("Nº: ", addr.group(2))


    if addr.group(3) == None:
      print("Sem nome de apartamento.")
    else:
      print("Apt: ", re.sub(',', '', addr.group(3).split(' ')[1]))
    
    if addr.group(4) == None:
      print("Sem nome de bairro.")
    else:
      print("Bairro: ", addr.group(4).split(',')[0])
    
    print("CEP: ", addr.group(5).split(' ')[1])
    print("Cidade: ", addr.group(6))
    print("Estado: ", addr.group(7))
```

```python
print(is_valid_address("Rua José da Silva, 346, Farol, CEP 57002-220. Maceió, AL."))
```

```python
print(is_valid_address("Rua da Consolação, 9999, ap 302, CEP 11022-202. São Paulo, SP."))
```

```python
print(is_valid_address("Avenida Atlântica, 420, apt 101, Copacabana, CEP 22011-101. Rio de Janeiro, RJ."))
```

```python
def search_url(candidate_url):
  pattern = "([https://]*(w+\.)?(\w\.?)+\.\D+(/\w+)?)+"
  
  search = re.compile(pattern).search(candidate_url)

  if not search:
    return False
  else:
    return True
```

```python
def print_url(txt_str):
  urls = []
  words = txt_str.split()
  for idx in range(len(words)):
    if search_url(words[idx]) is True:
      urls.append(words[idx])
  
  if len(urls) > 0:
    print("MATCH!\nUrls: ")
    for idx in range(len(urls)):      
      print(urls[idx])
  else:
    print("NO MATCH!\nThe string don't have urls.")
```

```python
def search_reference(string):  
  pattern = "(\D+,(\s([A-Z]\.)+)+,?)+\s(\(\d\d\d\d\))\.\s(\D+\s*)+\.\s(\D+)\."
  patt = re.compile(pattern)
  result = patt.match(string)
  if not result:
    print("NO MATCH: invalid format.")
  else:
    print("MATCH! Infos:")
    print('Autores: ' + result.group(1))
    print('Ano: ' + result.group(4))
    print('Revista: ' + result.group(5))
    print('Editora: ' + result.group(6))
```

```python
import string

def change_to_format(str_data):
  new_format = ""
  day = str_data.split('/')[0]
  month = str_data.split('/')[1]
  year = str(int(str_data.split('/')[2]) % 100)
  #print("Day: ", day)
  #print("Month: ", month)
  #print("Year: ", year)
  new_format += month + '-'
  new_format += day + '-'
  new_format += year
  #print("New formate data: ", new_format)
  return new_format
```

```python
change_to_format("22/12/2222")
```

```python
def search_data(candidate_data):
  pattern = "((\d\d){1}/{1}(\d\d){1}/{1}(\d\d\d\d){1})"
  compile_data = re.compile(pattern)
  result = compile_data.search(candidate_data)

  if not result:
    #print("NO MATCH!")
    return False
  else:
    #print("MATCH!")
    #print("Before: ", result.group(1))
    return True
```

```python
search_data("edvonaldo")
print("---")
search_data("22/12//")
print("---")
search_data("22/12/2222")
```

```python
def regex_data(txt_str):
  new_data = ""
  words = txt_str.split()
  for idx in range(len(words)):
    if search_data(words[idx]) is True:
      words[idx] = change_to_format(words[idx])
      new_data += str((words[idx]))
    else:
      new_data += str((words[idx]))
      new_data += " "
    
    

  print(new_data)
```

# Questão 7

### Removendo todas as palavras que contem números

```python
df_corpus['content'] = df_corpus['content'].replace(to_replace=r'\w*\d\w*', value='', regex=True)
print(df_corpus)
```

### Transformando todas as palavras em minúsculas

```python
for idx_txt in range(df_corpus['content'].shape[0]):
  df_corpus['content'][idx_txt] = df_corpus['content'][idx_txt].lower()
```

```python
df_corpus.head()
```

### Remover pontuação

```python
def remove_punctuation(txt):
  return re.sub('[%s]' % re.escape(string.punctuation), ' ', txt)
```

```python
df_corpus['content'] = df_corpus['content'].apply(remove_punctuation)

df_corpus.head()
```

### Tokenizando os textos em palavras

```python
cap_tokens = RegexpTokenizer("[\w][\w]+")

def tokenize_txt(txt):
  return cap_tokens.tokenize(txt)

token_texts = df_corpus['content'].apply(tokenize_txt)

print(token_texts.shape)
print(token_texts)
```

```python
# Criando um dicionário usando CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
x_token_text = cv.fit_transform(df_corpus['content'])
pd_vectorize = pd.DataFrame(x_token_text.toarray(), columns=cv.get_feature_names())
print("Shape: ", pd_vectorize.shape)
print(pd_vectorize)
```

### Palavras mais frequentes

```python
pd_word_high_freq = pd.DataFrame(columns=pd_vectorize.columns)

print(pd_word_high_freq.shape)
```

```python
pd_word_high_freq = pd_vectorize.sum(axis=0, skipna=True)

pd_word_high_freq = pd_word_high_freq.to_frame()
```

```python
pd_word_high_freq = pd_word_high_freq.sort_values(by=0, ascending=False)
print("/--- 10 palavras mais frequentes ---/")
print(pd_word_high_freq.head(10))
```
