## Helper function

```python
import re
```

```python
def search_pattern_in_string(pattern, string):
    '''
    Returns the string from the first match of 
    `pattern` in `string`.
    Returns "No match" if not found.
    '''
    result = re.compile(pattern).search(string)
    if not result:
        return "No match"
    else:
        return "Found pattern: " + result.group()
        
```

## Finding patterns using metacharacters

```python
print(search_pattern_in_string(".", "rlock Holmes")) # . Matches any single character
```

```python
print(search_pattern_in_string("^S", "Sherlock Holmes")) # ^ Beginning of string
print(search_pattern_in_string("^e", "Sherlock Holmes"))
print(search_pattern_in_string("^H", "Sherlock Holmes"))
```

```python
print(search_pattern_in_string("s$", "Sherlock Holmes")) # $ End of string
print(search_pattern_in_string("t$", "Sherlock Holmes"))
```

```python
print(search_pattern_in_string("w*", "Sherlock Holmes")) # empty string
print(search_pattern_in_string("S*", "Sherlock Holmes")) # * matches 0 or more characters
print(search_pattern_in_string("Sha*", "Sherlock Holmes"))
print(search_pattern_in_string("Swa*", "Sherlock Holmes"))
print(search_pattern_in_string("Sha*", "Shaaaaaaaerlock Holmes"))
```

```python
print(search_pattern_in_string("z*", "Sherlock Holmes")) # empty string
print(search_pattern_in_string("z+", "Sherlock Holmes")) # + must match 1 or more characters
print(search_pattern_in_string("S+", "Sherlock Holmes"))
print(search_pattern_in_string("Sha+", "Sherlock Holmes"))
```

```python
print(search_pattern_in_string("S3?h", "Sherlock Holmes")) # '3' is optional
print(search_pattern_in_string("S3?h", "S3herlock Holmes")) # '3' is optional
```

```python
print(search_pattern_in_string("2{1,3}", "221B Baker Street, London")) #{m,n} specify number of times character is matched between m and n times
print(search_pattern_in_string("2{3,4}", "221B Baker Street, London"))
```

```python
print(search_pattern_in_string("[ie]", "221B Baker Street, London")) # list characters to be matched
print(search_pattern_in_string("[f-i]", "221B Baker Street, London"))
print(search_pattern_in_string("[f-m]", "221B Baker Street, London"))
print(search_pattern_in_string("[fik]", "221B Baker Street, London"))
```

```python
print(search_pattern_in_string("z|k", "221B Baker Street, London")) # OR
```

```python
print(search_pattern_in_string(",", "221B Baker Street, London"))
```

```python
string = "Is there any other point to which you would wish to draw my attention?"
# print(search_pattern_in_string("?", string)) # would error
print(search_pattern_in_string("\?", string)) # error
```

```python
print(search_pattern_in_string("\s", "221B Baker Street, London")) # \s - matches any whitespace
print(search_pattern_in_string("\sBaker", "221B Baker Street, London")) 
print(search_pattern_in_string("\w", "221B Baker Street, London")) # \w - Matches any alphanumeric character; this is equivalent to the class [a-zA-Z0-9_]
print(search_pattern_in_string("\d", "221B Baker Street, London")) # \d - matches any numeric character. Equivalent to [0-9]
print(search_pattern_in_string("\D", "221B Baker Street, London")) # You may negate these by capitalizing. For example, \D matches anything not a digit
```

```python
print(search_pattern_in_string("[^0-9|B|\s]", "221B Baker Street, London")) # ^ inside [] means complement of set
```

```python
print(search_pattern_in_string("\d+B?\s\w+\s", "221B Baker Street, London")) # ^ inside [] means complement of set
```

```python
#Greedy repetitions:
print(search_pattern_in_string("Sher[a-zA-Z\s]*mes", "Sherlock Holmes")) # due to the * after the [], keeps finding a-zA-Z\s pattern until it is not found. than goes to 'mes'
```

```python
print(search_pattern_in_string("\d+\w* [A-Z]{1}[a-z]+ [A-Z]{1}[a-z]+", "221B Baker Street, London"))
```

## Groups

```python
pattern = "(\d+\w*) (([A-Z]{1}[a-z]+\s*)+), ([A-Z]{1}[a-z]+)"
p = re.compile(pattern)
m = p.search("221B George Teste Washington Street, London")
print(m.group(0)) # entire address
print(m.group(1)) # first part of address
print(m.group(2)) # second part of address
print(m.group(3)) # last word of second part of address
print(m.group(4)) # third part of address
```

```python
m.groups()
```

```python
pattern = "(\d+\w*) ((?:[A-Z]{1}[a-z]+\s*)+), ([A-Z]{1}[a-z]+)"
p = re.compile(pattern)
m = p.search("221B George Teste Washington Street, London")
print(m.group(0)) # entire address
print(m.group(1)) # first part of address
print(m.group(2)) # second part of address
print(m.group(3)) # third part of address
```

#Lookaheads/lookbehinds

```python
p = re.compile("[a-z]+(?=\d)") # Look ahead: (?=\d) checks to see if the character is followed by a number
if p.search('apple87'):
    print(p.search('apple87').group())
else:
    print('não achou')
if p.search('apple'):
    print(p.search('apple').group())
else:
    print('não achou')
```

```python
p = re.compile("(?<=[0-9][0-9])[a-z]+") # Look Behind: (?<=[aeiou]) checks to see if the character is preceded by a vowel
p.search('87apple').group()
```
